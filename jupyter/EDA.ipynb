#### ClienteId

```python
def limpiar_cliente_id(serie):
    # Formatos que hay: CLI_0572, CLI0023, cli-625 o solo el número. Lo demás se queda en NaN
    return serie.str.strip().str.extract(r'(?i)^(?:cli[_-]?)?(\d+)$')[0]
```

```python
# Salen la cabecera repetida, la fila de "840 registros" y una vacía. Las quitamos
antes = len(df_cl)
df_cl['ClienteId'] = ids
df_cl = df_cl.dropna(subset=['ClienteId'])
df_cl['ClienteId'] = df_cl['ClienteId'].astype(int)
print(f"Filas eliminadas por ClienteId no válido: {antes - len(df_cl)}")
```

#### Email

```python
def validar_email(serie):
    s = serie.str.strip().str.lower()
    # algo antes de la @, algo después y un punto en el dominio
    patron = r'^[^@\s]+@[^@\s]+\.[^@\s]+$'
    return s.where(s.str.match(patron, na=False))
```

```python
df_cl['Email'] = validar_email(df_cl['Email'])
```

#### Teléfono

```python
def limpiar_telefono(serie):
    # Quitamos lo que no sean números y nos quedamos con los 9 últimos, así se va el 34 o el 0034
    s = serie.str.replace(r'\D', '', regex=True).str[-9:]
    # Tiene que tener 9 dígitos y empezar por 6 o 7. Si no, a NaN
    valido = (s.str.len() == 9) & s.str[0].isin(['6', '7'])
    return s.where(valido)
```

```python
df_cl['Telefono'] = limpiar_telefono(df_cl['Telefono'])
```

#### Código postal

```python
def limpiar_cp(serie):
    s = serie.str.replace(r'\D', '', regex=True)
    # Los vacíos o con más de 5 dígitos no se pueden arreglar, a NaN
    s = s.where(s.str.len().between(1, 5))
    # A los de menos de 5 les faltan ceros delante, 8151 pasa a 08151
    return s.str.zfill(5)
```

```python
df_cl['CodigoPostal'] = limpiar_cp(df_cl['CodigoPostal'])
```

Se mantienen CP y Telefono como texto porque no son valores de operaciones.

#### Ingresos mensuales

```python
def limpiar_ingresos(serie):
    # Fuera €, EUR, espacios y textos (n/d, Sin ingresos). Dejamos números, coma, punto y el signo menos
    s = serie.str.replace(r'[^\d,.\-]', '', regex=True)
    # Si hay coma es el decimal y los puntos son de miles, como en 3.209,17
    con_coma = s.str.contains(',', na=False)
    s = s.where(~con_coma, s.str.replace('.', '', regex=False).str.replace(',', '.', regex=False))
    # Sin coma: un punto con 3 dígitos detrás es de miles (3.029), con 1 o 2 es decimal (1497.82)
    miles = s.str.match(r'^-?\d{1,3}(\.\d{3})+$', na=False)
    s = s.where(~miles, s.str.replace('.', '', regex=False))
    return pd.to_numeric(s, errors='coerce')
```

```python
df_cl['IngresosMensuales'] = limpiar_ingresos(df_cl['IngresosMensuales'])
```

Hay ingresos negativos y no tienen sentido como ingreso mensual. Se convierten a positivos, presumiblemente error de typo.

```python
def limpiar_aval(serie):
    s = serie.str.strip().str.lower()
    # Todas las formas de decir sí y no que hay en la columna
    mapa = {'true': True, 'si': True, 'sí': True, 's': True, '1': True,
            'false': False, 'no': False, 'n': False, '0': False}
    return s.map(mapa).astype('boolean')
```

```python
df_cl['TieneAval'] = limpiar_aval(df_cl['TieneAval'])
```

### FECHAS

```python
def limpiar_fecha(serie):
    s = serie.str.strip()
    # Formatos que hay: 2025-08-31, 2025/07/20, 24/03/2025, 05.05.2025, 12-09-25 y July 30, 2025
    formatos = ['%Y-%m-%d', '%Y/%m/%d', '%d/%m/%Y', '%d.%m.%Y', '%d-%m-%y', '%B %d, %Y']
    res = pd.Series(pd.NaT, index=s.index, dtype='datetime64[ns]')
    for f in formatos:
        res = res.fillna(pd.to_datetime(s, format=f, errors='coerce'))
    return res
```

```python
# Se comprueban las fechas inválidas

fechas = limpiar_fecha(df_cl['FechaEntrada'])
fallos = fechas.isna() & df_cl['FechaEntrada'].notna()
df_cl.loc[fallos, 'FechaEntrada'].value_counts()
```

### Acreedores

```python
def limpiar_acreedores(serie):
    # Separadores que hay: ; , / | y la palabra "y". Lo dejamos todo con "; "
    s = serie.str.strip().str.replace(r'\s+', ' ', regex=True)
    return s.str.replace(r'\s*[;,/|]\s*|\s+y\s+', '; ', regex=True)
```

```python
df_exp['Acreedores'] = limpiar_acreedores(df_exp['Acreedores'])
df_exp['Acreedores'].sample(5)
```

### Estado del expediente

```python
def limpiar_estado(serie):
    s = serie.str.strip().str.replace(r'\s+', ' ', regex=True).str.title()
    correcciones = {
        'Aceptado/En Negociacion': 'Aceptado / En Negociación',
        'Aceptado / En Negociacion': 'Aceptado / En Negociación',
        'Aceptado': 'Aceptado / En Negociación',
        'Activo': 'Plan Activo',
        'Estudio': 'En Estudio',
        'Cancelado': 'Cancelado Con Éxito',
        'Cancelado Con Exito': 'Cancelado Con Éxito',
        '-': np.nan,
        'N/D': np.nan,
    }
    return s.replace(correcciones)
```

```python
df_exp['EstadoExpediente'] = limpiar_estado(df_exp['EstadoExpediente'])
df_exp['EstadoExpediente'].value_counts(dropna=False)
```

```python
df_ofi = df_oficinas_raw.rename(columns={
    'nombre oficina': 'NombreOficina',
    'objetivo_conversion_%': 'ObjetivoConversionPct',
    'Fecha Apertura': 'FechaAperturaOficina',
})
```

```python
def limpiar_objetivo(serie):
    s = serie.str.replace('%', '', regex=False).str.replace(',', '.', regex=False).str.strip()
    n = pd.to_numeric(s, errors='coerce')
    # Hay valores como 0.30 y otros como 30. Lo dejamos todo en porcentaje
    return n.where(n > 1, n * 100).round(1)

df_ofi['NombreOficina'] = limpiar_oficina(df_ofi['NombreOficina'].str.strip().str.title())
df_ofi['Responsable'] = df_ofi['Responsable'].str.strip().str.title()
df_ofi['ObjetivoConversionPct'] = limpiar_objetivo(df_ofi['ObjetivoConversionPct'])
df_ofi['FechaAperturaOficina'] = limpiar_fecha(df_ofi['FechaAperturaOficina'])
```

```python
df_ofi = df_ofi.drop_duplicates()
df_ofi[df_ofi['NombreOficina'].duplicated(keep=False)]
```
