Unificamos formatos

```python
df_cl.sample(10)
```

```python
df_cl['ClienteId'] = df_cl['ClienteId'].str.lower().str.replace(r'^(cli_|cli-|cli)', '', regex=True)
```

```python
df_cl.sample(10)
```

```python
df_cl["ClienteId"].drop_duplicates().sort_values().tail(5)
```

```python
valido = df_cl["ClienteId"].str.contains(r"\d", na=False)   # True si tiene algún dígito
```

```python
df_cl[~valido]
```

```python
antes = len(df_cl)
df_cl = df_cl[valido]
print(f"Filas eliminadas por ClienteId no válido: {antes - len(df_cl)}")
```

```python
# Convierte la columna; los textos que no sean números pasarán a ser NaN
df_cl['ClienteId'] = pd.to_numeric(df_cl['ClienteId'], errors='coerce')
```

```python
df_cl.info()
```

```python
cols_texto = ['NombreCompleto', 'CanalCaptacion', 'DepartamentoOficina']
df_cl[cols_texto] = df_cl[cols_texto].apply(lambda col: col.str.title())
```

```python
df_cl['Email'] = df_cl['Email'].str.lower()
```
